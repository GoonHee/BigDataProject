```python
sns.countplot(x='Is_Fraud', data=df)

plt.title("Fraud vs Non-Fraud Transactions")
plt.show()
```

```python
plt.figure(figsize=(10,5))

sns.histplot(df['Transaction_Amount'], bins=50)

plt.title("Transaction Amount Distribution")
plt.show()
```

```python
sns.boxplot(x='Is_Fraud', y='Transaction_Amount', data=df)

plt.title("Fraud vs Transaction Amount")
plt.show()
```

```python
fraud_by_type = pd.crosstab(
    df['Transaction_Type'],
    df['Is_Fraud'],
    normalize='index'
)

print(fraud_by_type)
```

```python
fraud_by_type[1].sort_values().plot(
    kind='barh'
)
plt.title("Fraud Rate by Transaction Type")
plt.show()
```

```python
plt.figure(figsize=(12,6))

sns.countplot(x='Merchant_Category', hue='Is_Fraud', data=df)

plt.xticks(rotation=90)

plt.title("Fraud by Merchant Category")
plt.show()
```

```python
merchant_fraud = pd.crosstab(
    df['Merchant_Category'],
    df['Is_Fraud'],
    normalize='index'
)

merchant_fraud[1].sort_values().plot(
    kind='barh'
)
```

```python
device_fraud = pd.crosstab(
    df['Device_Type'],
    df['Is_Fraud'],
    normalize='index'
)

print(device_fraud)
```

```python
state_fraud = pd.crosstab(
    df['State'],
    df['Is_Fraud'],
    normalize='index'
)

state_fraud[1].sort_values(
    ascending=False
).head(10)
```

```python
state_fraud[1].sort_values(
    ascending=False
).head(10).plot(
    kind='bar'
)
```

```python
df['Transaction_Date'] = pd.to_datetime(
    df['Transaction_Date'],
    dayfirst=True
)
```

```python
df['Month'] = df['Transaction_Date'].dt.month
df['Day'] = df['Transaction_Date'].dt.day
```

```python
daily_fraud = df.groupby('Day')['Is_Fraud'].mean()

daily_fraud.plot()
plt.title("Daily Fraud Rate")
plt.show()
```

```python
df['Transaction_Time'] = pd.to_datetime(
    df['Transaction_Time']
).dt.hour
```

```python
hourly_fraud = df.groupby(
    'Transaction_Time'
)['Is_Fraud'].mean()

hourly_fraud.plot()
```

```python
numeric_cols = [
    'Age',
    'Transaction_Amount',
    'Account_Balance',
    'Is_Fraud'
]

corr = df[numeric_cols].corr()

sns.heatmap(
    corr,
    annot=True,
    cmap='coolwarm'
)
```
